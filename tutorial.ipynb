````python
# Helper T cell markers
helper_t_genes = ["CD3D", "CD4", "IL7R"]

result = interpret_marker_genes(helper_t_genes)
print(format_output(result))
````

## Example 5: Batch Processing

Analyze multiple cell types at once:

````python
from marker_interpreter import batch_interpret

# Multiple gene lists
gene_lists = [
    ["CD3D", "CD4", "IL7R"],      # Helper T cells
    ["CD19", "MS4A1", "CD79A"],   # B cells
    ["CD14", "LYZ", "S100A8"],    # Monocytes
]

results = batch_interpret(gene_lists)

# Display all results
for i, result in enumerate(results):
    print(f"\n=== Cell Population {i+1} ===")
    print(format_output(result))
````
